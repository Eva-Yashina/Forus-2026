Шаг 2. Найти сумму цифр после "запятой" и вывести на экран.

Шаг 3. Вывести последовательность для числа **из шага 2**, согласно гипотезе Коллатца, пока не будет достигнута единица. Последовательность необходимо сохранить в массив.

**Гипотеза Коллатца**

$f_{n} = \left\{\begin{array}. 3f_{n-1}+1, \text{если } f_{n-1} \text{ нечетное}\\f_{n-1} / 2,\text{если } f_{n-1} \text{ четное}\end{array}\right.$

Например, для числа 13 последовательность будет:

13 -> 40 -> 20 -> 10 -> 5 -> 16 -> 8 -> 4 -> 2 -> 1


**!Примечание**

При вводе некорректных данных на шаге 1. программа должна "переспрашивать", при этом не должно возникать ошибок.

```python
#Решение
while True:
  try:
    print("Введите число больше 8.0, но меньше 9.0 (не включительно).")
    num = float(input())

    if num > 8.0 and num < 9.0:

      print(f"Введённое число:{num}")

      num_str = str(num)
      fr = num_str.split('.')[1]
      sum_n = sum(int(i) for i in fr)
      print (f"Сумма цифр после запятой {sum_n}")

      sequence = [sum_n]

      while sum_n !=1:
        if sum_n % 2 == 0:
            sum_n = sum_n // 2
        else:
            sum_n = 3 * sum_n + 1
        sequence.append(sum_n)
      print(f"Последовательность: {sequence}")

      steps = len(sequence) -1
      print(f"Количество шагов в последовательности: {steps}")

      max_sum_n = max(sequence)
      print(f"Самое большое число в последовательности: {max_sum_n}")
      break

    else:
      print("Введите другое число!")

  except ValueError:
    print("Вводите только число!")
```
