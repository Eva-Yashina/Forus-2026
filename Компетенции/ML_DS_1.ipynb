#Задание пятое

Найдите минимум функции $f(x)$ градиентным методом, используя начальное приближение $x_0 = [2, 3]$:

$f(x, y) = sin(x) + cos(y) + ln(xy) \downarrow \min, x \in \mathbb{R}^2$

Используйте шаг градиентного метода, равный $0.1$

Для критерия остановки метода выберете условие:

$||x^k - x^{k+1}||_2 < 10^{-6}$.

Покажите, что найденная точка является локальным минимумом (приближённо).

Ответьте на вопросы, указанные ниже:
1. Сколько итераций метода потребовалось?
2. Какое значение целевой функции в найденной точке?
3. Что можно сказать о глобальной оптимальности найденной точки?

```python
#Решение здесь
import numpy as np

def f(x):
    return np.sin(x[0]) + np.cos(x[1]) + np.log(x[0] * x[1])

def grad(x):
    x1, x2 = x
    df_dx = np.cos(x1) + 1 / x1
    df_dy = -np.sin(x2) + 1 / x2
    return np.array([df_dx, df_dy])

alpha = 0.1
x = np.array([2.0, 3.0])

eps = 1e-6
max_iter = 10000

for i in range(max_iter):
    g = grad(x)
    x_new = x - alpha * g

    if np.linalg.norm(x_new - x) < eps:
        break

    x = x_new

g_final = grad(x)

x1, x2 = x
H = np.array([
    [-np.sin(x1) - 1/x1**2, 0],
    [0, -np.cos(x2) - 1/x2**2]
])

eigvals = np.linalg.eigvals(H)
```

```python
#Впишите ответы здесь
print('1. Количество итераций:', i + 1)
print('2. Значение функции:', f(x))
print('Координаты точки минимума:', x)
print('Градиент в точке:', g_final)
print('Гессиан:\n', H)
print('Собственные значения:', eigvals)

if np.all(eigvals > 0):
    print('3. Найденная точка является локальным минимумом')
else:
    print('3. Локальный минимум не подтвержден')

print('Глобальный минимум не гарантируется, так как функция невыпуклая')
```
